````python
import pandas as pd
# Specify the path to the file
file_path = '/content/drive/MyDrive/antigen_epitope/YUHAO/data/antigen_info_SI/1bvk_B_A_C_antigen_info_SI.tsv'

# Read the TSV file into a DataFrame
df = pd.read_csv(file_path, sep='\t')

# Display the first few rows of the DataFrame to check
print(df.head())
````

````python
import pandas as pd

# Assuming df is your DataFrame with the SI information

# Path to the PDB file
pdb_path = '/content/drive/MyDrive/antigen_epitope/YUHAO/data/antigen_pdbs/1bvk_B_A_C_antigen.pdb'

# Read the PDB file
with open(pdb_path, 'r') as file:
    pdb_lines = file.readlines()

# Function to format the B-factor in the PDB line
def format_b_factor(b_factor):
    return f"{b_factor:6.2f}"

# Update B-factors in the PDB file lines
updated_lines = []
for line in pdb_lines:
    if line.startswith("ATOM") or line.startswith("HETATM"):
        residue_number = int(line[22:26].strip())
        residue_name = line[17:20].strip()
        matching_rows = df[(df['ResID'] == residue_number) & (df['AA'] == residue_name)]
        if not matching_rows.empty:
            b_factor = matching_rows.iloc[0]['SI']
            new_line = line[:60] + format_b_factor(b_factor) + line[66:]
            updated_lines.append(new_line)
        else:
            updated_lines.append(line)
    else:
        updated_lines.append(line)

# Write the updated PDB file
with open(pdb_path.replace('.pdb', '_updated.pdb'), 'w') as file:
    file.writelines(updated_lines)

print("PDB file updated.")
````
