The following code extracts the Antigen and antibody chains.
Calculates the distance of their **atoms** between each other.

```python
import os
from Bio import PDB
import matplotlib.pyplot as plt

#Extract H and L chains from the antigen
def find_antigen_chains(pdb_file, antibody_heavy_chain, antibody_light_chain):
    parser = PDB.PDBParser(QUIET=True)
    structure = parser.get_structure("structure", pdb_file)

    antigen_chains = []
    for model in structure:
        for chain in model:
            if chain.id not in [antibody_heavy_chain, antibody_light_chain]:
                antigen_chains.append(chain.id)
    return antigen_chains

#Calculate which atoms are interacting between the ab and antigen chains
def find_interactions(pdb_file, antibody_chains, antigen_chains, distance_cutoff=4.0):
    parser = PDB.PDBParser(QUIET=True)
    structure = parser.get_structure('structure', pdb_file)

    antibody_atoms = []
    for chain in antibody_chains:
        antibody_atoms.extend([atom for atom in structure[0][chain].get_atoms()])

    interacting_antigen_atoms = set()

    for antigen_chain in antigen_chains:
        antigen_atoms = [atom for atom in structure[0][antigen_chain].get_atoms()]
        for antibody_atom in antibody_atoms:
            for antigen_atom in antigen_atoms:
                distance = antibody_atom - antigen_atom
                if distance <= distance_cutoff:
                    interacting_antigen_atoms.add(antigen_atom)

    return interacting_antigen_atoms

def main(folder_path, antibody_heavy_chain, antibody_light_chain):
    pdb_files = [f for f in os.listdir(folder_path) if f.endswith('.pdb')]
    interaction_counts = []

    for pdb_file in pdb_files:
        pdb_path = os.path.join(folder_path, pdb_file)
        antigen_chains = find_antigen_chains(pdb_path, antibody_heavy_chain, antibody_light_chain)
        interacting_atoms = find_interactions(pdb_path, [antibody_heavy_chain, antibody_light_chain], antigen_chains)
        interaction_counts.append(len(interacting_atoms))
        print(f"{pdb_file}: {len(interacting_atoms)} interacting atoms on antigen")

    return interaction_counts

if __name__ == "__main__":
    folder_path = "/content/drive/MyDrive/abdb_felipe/step0/pdb"
    antibody_heavy_chain = "H"
    antibody_light_chain = "L"

    interaction_counts = main(folder_path, antibody_heavy_chain, antibody_light_chain)

    plt.hist(interaction_counts, bins=10, edgecolor='black')
    plt.title('Distribution of Antigen-Antibody Interactions')
    plt.xlabel('Number of Interacting Atoms on Antigen')
    plt.ylabel('Frequency')
    plt.show()
```

This code loops through each **residue** in each antigen and calculates the amount of interacting residues per antigen using the distance between atoms calculated in the previous step.

```python
import os
from Bio import PDB
import matplotlib.pyplot as plt

def find_antigen_chains(pdb_file, antibody_heavy_chain, antibody_light_chain):
    parser = PDB.PDBParser(QUIET=True)
    structure = parser.get_structure("structure", pdb_file)

    antigen_chains = []
    for model in structure:
        for chain in model:
            if chain.id not in [antibody_heavy_chain, antibody_light_chain]:
                antigen_chains.append(chain.id)
    return antigen_chains

def find_interactions(pdb_file, antibody_chains, antigen_chains, distance_cutoff=4.0):
    parser = PDB.PDBParser(QUIET=True)
    structure = parser.get_structure('structure', pdb_file)

    antibody_atoms = []
    for chain in antibody_chains:
        antibody_atoms.extend([atom for atom in structure[0][chain].get_atoms()])

    interacting_antigen_residues = set()

    for antigen_chain in antigen_chains:
        antigen_residues = structure[0][antigen_chain].get_residues()
        for antigen_residue in antigen_residues:
            for antibody_atom in antibody_atoms:
                for antigen_atom in antigen_residue:
                    distance = antibody_atom - antigen_atom
                    if distance <= distance_cutoff:
                        interacting_antigen_residues.add(antigen_residue)
                        break

    return interacting_antigen_residues

def main(folder_path, antibody_heavy_chain, antibody_light_chain):
    pdb_files = [f for f in os.listdir(folder_path) if f.endswith('.pdb')]
    interaction_counts = []

    for pdb_file in pdb_files:
        try:
            pdb_path = os.path.join(folder_path, pdb_file)
            antigen_chains = find_antigen_chains(pdb_path, antibody_heavy_chain, antibody_light_chain)
            interacting_residues = find_interactions(pdb_path, [antibody_heavy_chain, antibody_light_chain], antigen_chains)
            interaction_counts.append(len(interacting_residues))
            print(f"{pdb_file}: {len(interacting_residues)} interacting residues on antigen")
        except Exception as e:
            print(f"Error processing {pdb_file}: {e}")

    return interaction_counts

if __name__ == "__main__":
    folder_path = "/content/drive/MyDrive/@antigen_epitope/abdb_v3/pdb"
    antibody_heavy_chain = "H"
    antibody_light_chain = "L"

    interaction_counts = main(folder_path, antibody_heavy_chain, antibody_light_chain)

    plt.hist(interaction_counts, bins=10, edgecolor='black')
    plt.title('Distribution of Antigen-Antibody Interactions')
    plt.xlabel('Number of Interacting Residues on Antigen')
    plt.ylabel('Frequency')
    plt.show()
```

```python
import os
from Bio import PDB

def find_antigen_chains(pdb_file, antibody_heavy_chain, antibody_light_chain):
    parser = PDB.PDBParser(QUIET=True)
    structure = parser.get_structure("structure", pdb_file)

    antigen_chains = []
    for model in structure:
        for chain in model:
            if chain.id not in [antibody_heavy_chain, antibody_light_chain]:
                antigen_chains.append(chain.id)
    return antigen_chains

def find_interactions_and_set_bfactor(pdb_file, antibody_chains, antigen_chains, output_dir, distance_cutoff=4.0):
    parser = PDB.PDBParser(QUIET=True)
    structure = parser.get_structure('structure', pdb_file)

    antibody_atoms = []
    for chain in antibody_chains:
        antibody_atoms.extend([atom for atom in structure[0][chain].get_atoms()])

    interacting_antigen_residues = set()

    # Set all B-factors to 0 initially
    for model in structure:
        for chain in model:
            for residue in chain:
                for atom in residue:
                    atom.set_bfactor(0.0)

    for antigen_chain in antigen_chains:
        antigen_residues = structure[0][antigen_chain].get_residues()
        for antigen_residue in antigen_residues:
            for antibody_atom in antibody_atoms:
                for antigen_atom in antigen_residue:
                    distance = antibody_atom - antigen_atom
                    if distance <= distance_cutoff:
                        interacting_antigen_residues.add(antigen_residue)
                        # Set B-factor to 1 for interacting residues
                        for atom in antigen_residue:
                            atom.set_bfactor(1.0)
                        break

    # Save the modified structure
    io = PDB.PDBIO()
    io.set_structure(structure)
    output_file = os.path.join(output_dir, os.path.basename(pdb_file).replace('.pdb', '_bfactor.pdb'))
    io.save(output_file)

    return interacting_antigen_residues, output_file

def main(input_folder, output_folder, antibody_heavy_chain, antibody_light_chain):
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

    pdb_files = [f for f in os.listdir(input_folder) if f.endswith('.pdb')]

    for pdb_file in pdb_files:
        try:
            pdb_path = os.path.join(input_folder, pdb_file)
            antigen_chains = find_antigen_chains(pdb_path, antibody_heavy_chain, antibody_light_chain)
            interacting_residues, output_file = find_interactions_and_set_bfactor(pdb_path, [antibody_heavy_chain, antibody_light_chain], antigen_chains, output_folder)
            print(f"{pdb_file}: {len(interacting_residues)} interacting residues on antigen. Output saved as {output_file}")
        except Exception as e:
            print(f"Error processing {pdb_file}: {e}")

if __name__ == "__main__":
    input_folder = "/content/drive/MyDrive/@antigen_epitope/abdb_v3/pdb"
    output_folder = "/content/drive/MyDrive/@antigen_epitope/abdb_v3/pdb_bfactor"
    antibody_heavy_chain = "H"
    antibody_light_chain = "L"

    main(input_folder, output_folder, antibody_heavy_chain, antibody_light_chain)
```

```python
import os
from collections import Counter

# Define the input and output folder paths
input_folder = "/content/drive/MyDrive/abdb_felipe/step0/pdb"
output_folder = "/content/drive/MyDrive/abdb_felipe/step0/pdb_bfactor"

# Function to count files in a folder
def count_files(folder):
    return len([f for f in os.listdir(folder) if os.path.isfile(os.path.join(folder, f))])

# Function to check for repeated files by their names (without extensions)
def find_repeated_files(folder):
    filenames = [os.path.splitext(f)[0] for f in os.listdir(folder) if os.path.isfile(os.path.join(folder, f))]
    filename_counts = Counter(filenames)
    repeated_files = [name for name, count in filename_counts.items() if count > 1]
    return repeated_files

# Count files in both folders
input_file_count = count_files(input_folder)
output_file_count = count_files(output_folder)

# Check for repeated files in both folders
input_repeated_files = find_repeated_files(input_folder)
output_repeated_files = find_repeated_files(output_folder)

# Print the results
print(f"Input Folder: {input_file_count} files")
if input_repeated_files:
    print(f"Repeated files in Input Folder: {input_repeated_files}")
else:
    print("No repeated files in Input Folder.")

print(f"Output Folder: {output_file_count} files")
if output_repeated_files:
    print(f"Repeated files in Output Folder: {output_repeated_files}")
else:
    print("No repeated files in Output Folder.")
```
