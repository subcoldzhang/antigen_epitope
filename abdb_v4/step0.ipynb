```python
import os
import shutil
from sklearn.model_selection import KFold

# Define the paths
base_path = '/content/drive/MyDrive/@antigen_epitope/abdb_v3/pdb'
output_path = '/content/drive/MyDrive/@antigen_epitope/abdb_v4'

# Get all pdb files
pdb_files = [f for f in os.listdir(base_path) if f.endswith('.pdb')]

# Number of splits for k-Fold Cross-Validation
n_splits = 5

# Create the KFold object
kf = KFold(n_splits=n_splits)

# Create output directories if they don't exist
if not os.path.exists(output_path):
    os.makedirs(output_path)

for i, (train_index, test_index) in enumerate(kf.split(pdb_files), start=1):
    train_files = [pdb_files[idx] for idx in train_index]
    test_files = [pdb_files[idx] for idx in test_index]

    # Print the number of files in train and test sets
    print(f"Fold {i}:")
    print(f"  Training files: {len(train_files)}")
    print(f"  Testing files: {len(test_files)}")

    # Create subdirectories for this fold
    train_dir = os.path.join(output_path, f'fold_{i}', 'data', 'abdb_data')
    test_dir = os.path.join(output_path, f'fold_{i}', 'data_test', 'abdb_data')
    os.makedirs(train_dir, exist_ok=True)
    os.makedirs(test_dir, exist_ok=True)

    # Copy files to the respective directories
    for file in train_files:
        shutil.copy(os.path.join(base_path, file), os.path.join(train_dir, file))

    for file in test_files:
        shutil.copy(os.path.join(base_path, file), os.path.join(test_dir, file))
```
